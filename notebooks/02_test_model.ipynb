```python
import pandas as pd
from xgboost import XGBClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import roc_auc_score, classification_report

# 1. Khởi tạo một vỏ mô hình rỗng
loaded_model = XGBClassifier()

# 2. Nạp trọng số từ thư mục models
loaded_model.load_model('../models/xgboost_risk_model.json')

# Tạo dữ liệu cho 1 sinh viên với đúng 10 features đã train
thong_tin_test = pd.DataFrame({
    'num_of_prev_attempts': [0],          
    'studied_credits': [80],             
    'Avg_Coursework_Score': [85.0],      
    'Total_late_days': [0],              
    'Submitted_assignments': [10],        
    'First_Assessment_Score': [90.0],    
    'Total_clicks': [800],               
    'Active_days': [25],                 
    'Forum_clicks': [30],                
    'Click_Momentum': [1.5]              
})

# Gọi mô hình để dự báo
nhan = loaded_model.predict(thong_tin_test)[0]
xac_suat = loaded_model.predict_proba(thong_tin_test)[0][1]

print("--- KẾT QUẢ DỰ BÁO ---")
print(f"Nhãn dự báo (0-An toàn, 1-Rủi ro): {nhan}")
print(f"Xác suất rớt môn: {xac_suat * 100:.2f}%")

```
